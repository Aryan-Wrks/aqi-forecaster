```python
import pandas as pd

old = pd.read_csv("../data/city_day.csv", parse_dates=["Date"])
print("city_day:", old.shape)
print(old.columns.tolist())
old.head()
```

```python
new = pd.read_csv("../data/aqi.csv")
print("aqi:", new.shape)
print(new.columns.tolist())
new.head()
```

```python
patna = old[old["City"] == "Patna"].sort_values("Date").set_index("Date")
print(patna.shape)
print(patna.index.min(), "to", patna.index.max())
patna.head()
```

```python
new["date"] = pd.to_datetime(new["date"], dayfirst=True)
print(new["date"].min(), "to", new["date"].max())
print(new["area"].nunique(), "areas")

patna_new = new[new["area"].str.contains("Patna", case=False, na=False)].sort_values("date")
print(patna_new.shape)
print(patna_new["area"].unique())
print(patna_new["date"].min(), "to", patna_new["date"].max())
patna_new.head()
```

```python
print(patna["AQI"].notna().sum(), "of", len(patna), "days have AQI")
print(patna["AQI"].first_valid_index(), "to", patna["AQI"].last_valid_index())
```

```python
print(patna_new["date"].duplicated().sum(), "duplicate dates")
full = pd.date_range(patna_new["date"].min(), patna_new["date"].max())
print(len(full) - patna_new["date"].nunique(), "missing days")
```

```python
a = patna_new.drop_duplicates("date").set_index("date")["aqi_value"].rename("bulletin_aqi")
b = patna["AQI"].rename("kaggle_aqi")
both = pd.concat([a, b], axis=1, join="inner").dropna()

print(len(both), "overlapping days")
print(both.corr())
print("Average difference:", (both["bulletin_aqi"] - both["kaggle_aqi"]).abs().mean())
both.head()
```

```python
patna_new = (new[new["area"].str.strip().str.lower() == "patna"]
             .sort_values("date")
             .set_index("date"))

print(patna_new.shape)
print(patna_new.index.min(), "to", patna_new.index.max())
print(patna_new.index.duplicated().sum(), "duplicate dates")

full = pd.date_range(patna_new.index.min(), patna_new.index.max())
missing = full.difference(patna_new.index)
print(len(missing), "missing days")
print(missing[:10])

print(patna_new["aqi_value"].describe())
print(patna_new["prominent_pollutants"].value_counts().head())
```

```python
aqi = patna_new[["aqi_value", "number_of_monitoring_stations", "prominent_pollutants"]].copy()

full_index = pd.date_range(aqi.index.min(), aqi.index.max(), freq="D")
aqi = aqi.reindex(full_index)
aqi.index.name = "date"

print(len(aqi), "days in calendar")
print(aqi["aqi_value"].isna().sum(), "missing AQI values")
print(aqi.loc["2024-12-29":"2025-01-04", "aqi_value"])
```

```python
aqi["was_filled"] = aqi["aqi_value"].isna()      # remember which days we filled

aqi["aqi_value"] = aqi["aqi_value"].interpolate(limit=3)
aqi["number_of_monitoring_stations"] = aqi["number_of_monitoring_stations"].ffill(limit=3)
aqi["prominent_pollutants"] = aqi["prominent_pollutants"].ffill(limit=3)

print(aqi.isna().sum())
print(aqi["was_filled"].sum(), "day(s) filled")
```

Cell A: missing share by year

```python
s = patna["AQI"]
print(s.isna().groupby(s.index.year).mean().round(2))
```

Cell B: the longest gaps

```python
is_na = s.isna()
run_id = (is_na != is_na.shift()).cumsum()

gaps = (pd.DataFrame({"na": is_na, "run": run_id})
        .query("na")
        .reset_index()
        .groupby("run")
        .agg(start=("Date", "min"), end=("Date", "max"), days=("Date", "size")))

print(gaps.sort_values("days", ascending=False).head(5))
print(len(gaps), "separate gaps in total")
```
